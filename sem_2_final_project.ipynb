```python
import requests
import pandas as pd
```

```python
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"

response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
data = response.json()

print(type(data))
```

```python
repositories = data["items"]

print("Number of repositories:", len(repositories))
```

```python
df = pd.DataFrame(repositories)

print(df.shape)

df.head()
```

```python
required_columns = [
    "name",
    "owner",
    "language",
    "stargazers_count",
    "forks_count",
    "watchers_count",
    "open_issues_count",
    "created_at",
    "updated_at",
    "license"
]

df = df[required_columns]

df.head()
```

```python
df["owner"] = [
    repo.get("owner", {}).get("login")
    for repo in repositories
]

df["license"] = [
    repo.get("license", {}).get("name") if repo.get("license") else None
    for repo in repositories
]

df.head()
```

```python
df.isnull().sum()
```

```python
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("No License")

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])

print(df[["created_at", "updated_at"]].dtypes)
```

```python
df = df.rename(columns={
    "stargazers_count": "stars",
    "forks_count": "forks",
    "watchers_count": "watchers",
    "open_issues_count": "open_issues",
    "created_at": "created_date",
    "updated_at": "updated_date"
})

print(df.columns.tolist())
```

```python
df.to_csv("github_projects.csv", index=False)
print("Dataset saved successfully as github_projects.csv")
```

```python
import sqlite3

conn = sqlite3.connect("github_projects.db")

df_sql = pd.read_csv("github_projects.csv")

print("Database connected successfully.")
```

```python
df_sql.to_sql("Repositories", conn, if_exists="replace", index=False)

print("Repositories table created successfully.")
```

```python
query = """
SELECT name, owner, language, stars, forks
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT
    COUNT(*) AS total_repositories,
    AVG(stars) AS average_stars
FROM Repositories;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT
    language,
    COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repository_count DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
import matplotlib.pyplot as plt

top_repositories = pd.read_sql_query("""
SELECT name, stars
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
""", conn)

plt.figure(figsize=(12, 6))
plt.barh(top_repositories["name"], top_repositories["stars"])
plt.xlabel("Stars")
plt.ylabel("Repository")
plt.title("Top 10 Most Popular GitHub Repositories")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
creation_trends = pd.read_sql_query("""
SELECT
    strftime('%Y', created_date) AS year,
    COUNT(*) AS repository_count
FROM Repositories
GROUP BY year
ORDER BY year;
""", conn)

plt.figure(figsize=(10, 6))
plt.plot(
    creation_trends["year"],
    creation_trends["repository_count"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Number of Repositories")
plt.title("GitHub Repository Creation Trends Over Time")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

*# 1. Why is it important to verify data collected from public APIs?*

It is important to verify data collected from public APIs because the data may contain missing, outdated, duplicated, or inaccurate information. Checking the data helps ensure that it is reliable and suitable for analysis. This reduces the risk of producing incorrect results and making wrong decisions based on unreliable data.


*# 2. Why should data analysts document the source of their data?*

Data analysts should document the source of their data to make the analysis transparent and reproducible. Recording the source allows others to understand where the data came from, verify its reliability, and repeat the analysis if needed. It also helps analysts track changes in the data source over time.


*# 3. How can missing or inaccurate data affect data analysis and decision-making?*

Missing or inaccurate data can lead to incorrect analysis results and misleading conclusions. It can affect calculations, trends, and comparisons, making it difficult to identify the real patterns in the data. As a result, decisions based on unreliable data may be ineffective or incorrect. Therefore, data should be checked and cleaned before analysis.
